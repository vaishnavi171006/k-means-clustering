```python
from sklearn.cluster import KMeans

X = df[['Annual Income (k$)', 'Spending Score (1-100)']]

cluster = KMeans(
    init='k-means++',
    n_clusters=4,
    random_state=42,
    n_init=10
)

df['Cluster_No'] = cluster.fit_predict(X)

df.head()
```

```python
import matplotlib.pyplot as plt

plt.style.use('dark_background')

fig = plt.figure(figsize=(20, 15))
ax = fig.add_subplot(111, projection='3d')

# Cluster 0
ax.scatter3D(
    df['Age'][df['Cluster_No'] == 0],
    df['Annual Income (k$)'][df['Cluster_No'] == 0],
    df['Spending Score (1-100)'][df['Cluster_No'] == 0],
    c='red', s=50
)

# Cluster 1
ax.scatter3D(
    df['Age'][df['Cluster_No'] == 1],
    df['Annual Income (k$)'][df['Cluster_No'] == 1],
    df['Spending Score (1-100)'][df['Cluster_No'] == 1],
    c='green', s=50
)

# Cluster 2
ax.scatter3D(
    df['Age'][df['Cluster_No'] == 2],
    df['Annual Income (k$)'][df['Cluster_No'] == 2],
    df['Spending Score (1-100)'][df['Cluster_No'] == 2],
    c='yellow', s=50
)

# Cluster 3
ax.scatter3D(
    df['Age'][df['Cluster_No'] == 3],
    df['Annual Income (k$)'][df['Cluster_No'] == 3],
    df['Spending Score (1-100)'][df['Cluster_No'] == 3],
    c='blue', s=50
)

ax.set_xlabel('Age')
ax.set_ylabel('Annual Income (k$)')
ax.set_zlabel('Spending Score (1-100)')

ax.view_init(elev=10, azim=170)
plt.title('3D Scatter Plot of Customer Clusters')
plt.show()
```
